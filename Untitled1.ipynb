```python
import torch
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig
)

MODEL_ID = "Qwen/Qwen2.5-3B-Instruct"

quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

llm_model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    quantization_config=quant_config,
    device_map="auto"
)

print("Model loaded successfully!")
print("Device:", llm_model.device)
```

```python
import re
import numpy as np
import pandas as pd
from androguard.core.apk import APK

def extract_apk_features(apk_path, raw_feature_list):
    """
    Parses an APK file and counts occurrences of syscalls and binder APIs
    to populate the raw 470-feature vector.
    """
    # Initialize counts for all expected raw features
    feature_counts = {feature: 0 for feature in raw_feature_list}

    try:
        # Load APK
        apk_obj = APK(apk_path)

        # Iterate through DEX files to find strings/method calls
        for dex in apk_obj.get_all_dex():
            # We search within the string pool and classes for matches
            for string_item in dex.get_strings():
                # Decode if it's a bytes object
                if isinstance(string_item, bytes):
                    try:
                        string_item = string_item.decode('utf-8', errors='ignore')
                    except Exception:
                        string_item = str(string_item)
                elif not isinstance(string_item, str):
                    string_item = str(string_item)

                for feature in raw_feature_list:
                    # Standardize feature matching (clean underscores or look for exact substrings)
                    clean_feature = feature.replace("_", "").lower()
                    if clean_feature and clean_feature in string_item.lower():
                        feature_counts[feature] += 1

            # Also scan method calls / API instructions
            for method in dex.get_methods():
                method_name = method.get_name()
                if isinstance(method_name, bytes):
                    try:
                        method_name = method_name.decode('utf-8', errors='ignore')
                    except Exception:
                        method_name = str(method_name)
                elif not isinstance(method_name, str):
                    method_name = str(method_name)

                for feature in raw_feature_list:
                    clean_feature = feature.replace("_", "").lower()
                    if clean_feature and clean_feature in method_name.lower():
                        feature_counts[feature] += 1

    except Exception as e:
        print(f"Error processing {apk_path}: {e}")

    # Convert dictionary to ordered feature array matching selector input exactly
    raw_vector = np.array([feature_counts[feature] for feature in raw_feature_list], dtype=np.float64)
    return raw_vector.reshape(1, -1)
```

```python
import logging
from collections import Counter

logging.disable(logging.CRITICAL)

invoke_counts = Counter()
methods_with_code = 0
instructions_checked = 0
classes_checked = 0

for cls in dex.get_classes():
    classes_checked += 1

    for method in cls.get_methods():
        code = method.get_code()

        if code is None:
            continue

        methods_with_code += 1

        for instruction in code.get_bc().get_instructions():
            instructions_checked += 1
            opcode_name = instruction.get_name()

            if opcode_name.startswith("invoke-"):
                invoke_counts[opcode_name] += 1

print("Classes checked:", classes_checked)
print("Methods with code:", methods_with_code)
print("Instructions checked:", instructions_checked)

print("\nInvocation counts by opcode:")
for opcode, count in sorted(invoke_counts.items()):
    print(f"{opcode}: {count}")

print("\nTotal invoke instructions:", sum(invoke_counts.values()))
```

```python
import logging
from collections import Counter

logging.disable(logging.CRITICAL)

called_methods = Counter()

for cls in dex.get_classes():
    for method in cls.get_methods():
        code = method.get_code()

        if code is None:
            continue

        for instruction in code.get_bc().get_instructions():
            if instruction.get_name().startswith("invoke-"):
                output = instruction.get_output()

                # Extract method references from the instruction
                if "->" in output:
                    called_methods[output.strip()] += 1

print("Unique referenced-method strings:", len(called_methods))
print("\nTop 40 referenced methods:")
for method_name, count in called_methods.most_common(40):
    print(f"{count:6}  {method_name}")
```

```python
import logging
import re
from collections import Counter

logging.disable(logging.CRITICAL)

called_methods = Counter()

for cls in dex.get_classes():
    for method in cls.get_methods():
        code = method.get_code()

        if code is None:
            continue

        for instruction in code.get_bc().get_instructions():
            if not instruction.get_name().startswith("invoke-"):
                continue

            output = instruction.get_output()

            # Extract the DEX method signature, excluding register names
            match = re.search(
                r"(L[^;]+;->[^(\s]+\(.*)",
                output
            )

            if match:
                signature = match.group(1).strip()
                called_methods[signature] += 1

print("Unique normalized method signatures:", len(called_methods))

print("\nTop 30 method signatures:")
for signature, count in called_methods.most_common(30):
    print(f"{count:6}  {signature}")
```

```python
import re

api_patterns = {
    "network": r"Ljava/net/|Lokhttp3/|Lretrofit2/|Landroid/webkit/",
    "file_io": r"Ljava/io/|Ljava/nio/file/|Landroid/system/",
    "process_execution": r"Ljava/lang/Runtime;|Ljava/lang/ProcessBuilder;",
    "reflection": r"Ljava/lang/reflect/|Ljava/lang/Class;->forName",
    "dynamic_loading": r"Ldalvik/system/|LDalvik/system/",
    "sms_telephony": r"Landroid/telephony/SmsManager;|Landroid/telephony/TelephonyManager;",
    "accessibility": r"Landroid/accessibilityservice/|Landroid/view/accessibility/",
    "crypto": r"Ljavax/crypto/|Ljava/security/MessageDigest;",
    "binder_ipc": r"Landroid/os/IBinder;|Landroid/os/Binder;|Landroid/os/Parcel;",
}

print("Static API reference counts by category:\n")

for category, pattern in api_patterns.items():
    matches = {
        signature: count
        for signature, count in called_methods.items()
        if re.search(pattern, signature)
    }

    print(
        f"{category:20} "
        f"references={sum(matches.values()):6} "
        f"unique_methods={len(matches):5}"
    )
```
