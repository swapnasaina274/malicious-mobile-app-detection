```python
# ============================================================
# PHASE 3 - CELL 11
# PADDLEOCR-VL TRANSFORMERS COMPATIBILITY FIX
# ============================================================

import copy
import torch

from transformers import (
    AutoConfig,
    AutoProcessor,
    AutoModelForImageTextToText
)

MODEL_ID = "PaddlePaddle/PaddleOCR-VL-1.6"
REVISION = "c5630abae1d9"

print("================================")
print("BUILDING COMPATIBILITY CONFIG")
print("================================")

# ------------------------------------------------------------
# 1. Load the original configuration
# ------------------------------------------------------------

print("Loading original configuration...")

ocr_config = AutoConfig.from_pretrained(
    MODEL_ID,
    revision=REVISION,
    trust_remote_code=True
)

print("✓ Original configuration loaded")

# ------------------------------------------------------------
# 2. Inspect vision configuration
# ------------------------------------------------------------

vision_config = ocr_config.vision_config

print("\nVision config class:",
      type(vision_config).__name__)

# ------------------------------------------------------------
# 3. Add the RoPE parameters expected by the current
#    Transformers PaddleOCR-VL implementation.
#
#    The model's top-level configuration already specifies
#    the required values.
# ------------------------------------------------------------

top_level_rope = getattr(
    ocr_config,
    "rope_scaling",
    None
)

if top_level_rope is None:

    raise RuntimeError(
        "Expected top-level rope_scaling configuration "
        "was not found."
    )

print("\nOriginal top-level RoPE configuration:")
print(top_level_rope)

# PaddleOCR-VL vision implementation expects rope_parameters.
vision_config.rope_parameters = {
    "rope_type": "default",
    "mrope_section": top_level_rope.get(
        "mrope_section",
        [16, 24, 24]
    )
}

print("\nInjected vision RoPE configuration:")
print(vision_config.rope_parameters)

# ------------------------------------------------------------
# 4. Verify the configuration
# ------------------------------------------------------------

if hasattr(vision_config, "rope_parameters"):

    print("\n✓ Compatibility attribute successfully created")

else:

    raise RuntimeError(
        "Compatibility attribute could not be created."
    )

print("\n================================")
print("COMPATIBILITY CONFIG READY")
print("================================")
```

```python
# ============================================================
# PHASE 3 - CELL 23
# PADDLEOCR-VL-1.6 SCREENSHOT ANALYSIS
# ============================================================

import torch

print("================================")
print("PADDLEOCR-VL-1.6 ANALYSIS")
print("================================")

prompt = """
Analyze this Android application screenshot for security analysis.

Extract the following:

1. ALL visible text exactly as readable as possible.
2. Important UI elements such as buttons, menus, dialogs, input fields,
   permission prompts, warnings, navigation elements, and icons.
3. Security-related indicators such as permissions, authentication,
   account information, device access, storage access, network actions,
   or warnings.
4. Potentially suspicious indicators such as deceptive messages,
   unexpected permission requests, fake system dialogs, urgent requests,
   credential requests, financial requests, or unusual UI behavior.
5. Observable user actions or application actions suggested by the screen.

Do not assume that an indicator is malicious.
Only report what is visually supported by the screenshot.

Return a structured, concise analysis.
"""

print("Preparing model input...")

messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "image",
                "image": screenshot
            },
            {
                "type": "text",
                "text": prompt
            }
        ]
    }
]

inputs = ocr_processor.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
    return_dict=True
)

# Move tensors to the model's active device
inputs = {
    k: v.to(ocr_model.device)
    if hasattr(v, "to")
    else v
    for k, v in inputs.items()
}

print("✓ Input prepared")

print("\nRunning PaddleOCR-VL-1.6...")
print("This may take a little while.")

with torch.inference_mode():

    generated_ids = ocr_model.generate(
        **inputs,
        max_new_tokens=1024,
        do_sample=False
    )

# Remove prompt tokens
generated_ids_trimmed = [
    out_ids[len(in_ids):]
    for in_ids, out_ids in zip(
        inputs["input_ids"],
        generated_ids
    )
]

output_text = ocr_processor.batch_decode(
    generated_ids_trimmed,
    skip_special_tokens=True
)[0]

print("\n================================")
print("PADDLEOCR-VL OUTPUT")
print("================================")

print(output_text)

print("\n================================")
print("ANALYSIS COMPLETE")
print("================================")
```

```python
# Classify extracted OCR text into useful evidence categories

SECURITY_TERMS = [
    "permission", "camera", "microphone", "location",
    "contacts", "sms", "password", "passcode", "otp",
    "verification code", "login", "sign in", "authenticate",
    "bank", "banking", "upi", "credit card", "debit card",
    "payment", "transaction", "privacy", "tracking",
    "accessibility", "admin", "install", "download"
]

STATE_TERMS = [
    "invalid", "error", "failed", "warning",
    "not appear", "does not appear", "json",
    "exception", "denied", "blocked"
]

def contains_term(text, terms):
    text_lower = text.lower()
    return any(term in text_lower for term in terms)

security_evidence = []
state_evidence = []
general_text = []

for item in ocr_items:
    text = item["text"]

    if contains_term(text, SECURITY_TERMS):
        security_evidence.append(item)

    elif contains_term(text, STATE_TERMS):
        state_evidence.append(item)

    else:
        general_text.append(item)

classified_evidence = {
    "security_relevant": security_evidence,
    "application_state": state_evidence,
    "general_text": general_text
}

print("===== OCR EVIDENCE CLASSIFICATION =====")
print("Security-relevant :", len(security_evidence))
print("Application-state :", len(state_evidence))
print("General text      :", len(general_text))

print("\n--- Security-relevant ---")
for item in security_evidence:
    print(item["text"])

print("\n--- Application-state ---")
for item in state_evidence:
    print(item["text"])
```

```python
# Convert ANY NumPy/Python numeric types recursively into JSON-safe types

import numpy as np

def make_json_safe(obj):
    if isinstance(obj, dict):
        return {
            str(k): make_json_safe(v)
            for k, v in obj.items()
        }

    if isinstance(obj, (list, tuple)):
        return [
            make_json_safe(v)
            for v in obj
        ]

    if isinstance(obj, np.integer):
        return int(obj)

    if isinstance(obj, np.floating):
        return float(obj)

    if isinstance(obj, np.ndarray):
        return obj.tolist()

    if isinstance(obj, np.bool_):
        return bool(obj)

    return obj


final_phase3 = {
    "source": screenshot_evidence["source"],
    "fast_ocr": {
        "engine": "EasyOCR",
        "inference_time_seconds": elapsed,
        "detected_regions": len(ocr_items),
        "items": ocr_items
    },
    "classified_evidence": {
        "security_relevant": security_evidence,
        "application_state": state_evidence,
        "general_text": general_text
    },
    "advanced_vlm": {
        "model": "PaddleOCR-VL-1.6",
        "available_as": "advanced_visual_evidence_fallback",
        "used": True
    }
}

# Make the COMPLETE package JSON-safe
final_phase3 = make_json_safe(final_phase3)

FINAL_PHASE3_PATH = "/content/final_phase3_evidence.json"

with open(FINAL_PHASE3_PATH, "w", encoding="utf-8") as f:
    json.dump(
        final_phase3,
        f,
        indent=2,
        ensure_ascii=False
    )

print("===== PHASE 3 OPTIMIZED =====")
print("Fast OCR time     :", final_phase3["fast_ocr"]["inference_time_seconds"], "seconds")
print("OCR regions       :", final_phase3["fast_ocr"]["detected_regions"])
print("Security evidence :", len(final_phase3["classified_evidence"]["security_relevant"]))
print("Application state :", len(final_phase3["classified_evidence"]["application_state"]))
print("General text      :", len(final_phase3["classified_evidence"]["general_text"]))
print("Saved             :", FINAL_PHASE3_PATH)
```
